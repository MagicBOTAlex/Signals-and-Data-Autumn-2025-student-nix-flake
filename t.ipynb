```python
from uncertainties import ufloat
# inputs (± are 1σ)
r1 = ufloat(0.038, 0.001)      # m
r2 = ufloat(0.052, 0.001)      # m
g  = 9.82                      # m/s^2
m_tape = ufloat(0.35847, 0.001/1000)  # kg  (0.1 g)
m_samlet = ufloat(0.55706, 0.001/1000)    # kg
r  = ufloat(0.052, 0.001)      # m
alpha = ufloat(12.962, 0.165)  # rad/s^2
katete = ufloat(0.097, 0.001)  # m
hypotenuse = ufloat(0.67, 0.001) # m

i_tape   = 2*(m_tape/2)*(r1**2 + r2**2)
i_samlet = (r*(m_samlet*g*(katete/hypotenuse) - m_samlet*(r*alpha)))/alpha
iolab    = i_samlet - i_tape

print("I_tape   =", i_tape)    # kg·m^2
print("I_samlet =", i_samlet)  # kg·m^2
print("I_olab   =", iolab)     # kg·m^2
```
